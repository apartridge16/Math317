## Problem 1

Implement the Pollard $p-1$ factorization algorithm.  Call your function 'pollard_p_1_factorization( )'.

````python
def pollard_p_1_factorization(N):
    a = 2
    for j in range(a, N):
        a = pow(a, j, N)
        g = gcd(a - 1, N)

        if 1 < g < N:
            return g

    return None
    
````

## Problem 2

Implement the Fermat factorization algorithm.  Call your function 'fermat_factorization( )'.

````python
def fermat_factorization(N):
    a = ceil(sqrt(N))

    while True:
        B = a^2 - N
        
        if is_square(B):
            b = sqrt(B)
            
            return (a - b, a + b)
            
        a += 1
    
````

````python
def pseudo_test(n, b):
    if gcd(b, n) != 1:
        return False
    
    if pow(b, n - 1, n) != 1:
        return False
    
    return True
````

## Problem 4

Code the strong pseudoprime test, and call your function `strong_pseudo_test(n,b)`.  Your function should take $n$ and $b$ as inputs, and return True if $n$ passes all $3$ tests with the base $b$ and False otherwise.

````python
def strong_pseudo_test(n,b):
    if not pseudo_test(n,b):
        return False

    q = n - 1
    k = 0
    while q % 2 == 0:
        q = q // 2
        k += 1

    if pow(b, q, n) == 1:
        return True

    for i in range(k):
        if pow(b, (2**i) * q, n) == n-1:
            return True
    
    return False
````

````python
def miller_rabin(n, k):
    if n % 2 == 0:
        return False
    
    for i in range(k):
        b = randint(2, n - 2)
        if not strong_pseudo_test(n, b):
            return False
    
    return True


while True:
    n = randint(10^499, 10^500 - 1)
    
    if miller_rabin(n, 10):
        print(n)
        break
````

## Problem 8

In this problem you will write functions to convert a string to an integer and vice versa.  Each character has an ascii value between 0 and 127 inclusively, so you can think of a string as an integer in the base 128.  I've written the encode function for you and you just need to write the decode function.

````python
def encode(a_string):
    return sum([ord(a_string[i])*128^i for i in range(len(a_string))])
````

````python
def decode(n):
    result = ""

    while n > 0:
        char_code = n % 128
        result += chr(char_code)
        n //= 128

    return result
````
