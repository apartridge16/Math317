and if yes, find its multiplicative inverse.

````python
import math
def ext_euclidean(a, b):

    s2 = 1
    s1 = 0

    t2 = 0
    t1 = 1

    r2 = a
    r1 = b

    while r1 != 0:

        q = math.floor(r2 / r1)

        r = r2 - q * r1
        r2 = r1
        r1 = r

        s = s2 - q * s1
        s2 = s1
        s1 = s

        t = t2 - q * t1
        t2 = t1
        t1 = t

    g = r2
    s_coeff = s2
    t_coeff = t2

    return [g, s_coeff, t_coeff]


print(ext_euclidean(104958729304785078924350987312450987103942587,
      143209587120394750191345))
````

````python
def fast_exp(n, k, m):
    k_bin = bin(k)[2:]
    rev = str(k_bin)[::-1]

    b = 1
    for i in range(len(rev)):
        if (rev[i] == '1'):
            power = pow(2, i)
            b *= pow(n, power, m)

    b = b % m

    return b   



print("Answer:", fast_exp(5, 3389, 7))
````

You are breaking a Diffie-Hellman key exchange.  The public values are $p = 71558387$ and $g=65537$. (You should be able to do this with a brute-force computation of the discrete log as we did in class with small numbers.)

````python
p = 71558387
g = 65537
A = 36858688   # g^a mod p
B = 11488428   # g^b mod p

value = 1
a = 1
b = 1
S = 1

for i in range(p):

    value = pow(g, i, p)

    if (value == A):
        a = i
        S = pow(B, a, p)
        break
    elif (value == B):
        b = i
        S = pow(A, b, p)
        break

print("Secret:", S)


````

## Problem 6

Write code that allow you to find Carmichael numbers (Fermat pseudoprimes) up to a bound.  Write as many functions as you wish.  You can use functions from previous homework assignments, such as your own `prime_range()` function.  Your function `fast_exp()` from Problem 3 of this current homework assignment will probably be handy as well.  How many can you find?  What is the largest one you can find? 

````python
def prime_range(n):

    sieve = [True] * (n + 1)

    sieve[0] = False
    sieve[1] = False

    p = 2
    while p * p <= n:

        if sieve[p] == True:

            multiple = p * p
            while multiple <= n:
                sieve[multiple] = False
                multiple += p

        p += 1

    primes = []
    i = 2
    while i <= n:
        if sieve[i] == True:
            primes.append(i)
        i += 1

    return primes
````

````python
def gcd(a, b):
    r = 1
    while (r >= 0 and b != 0):
        r = a % b
        if r == 0:
            break
        a = b
        b = r

    return b
````

````python
def is_carmichael(c_number, primes):

    if c_number in primes:
        return False

    for i in range(2, c_number):
        if (gcd(i, c_number) == 1):
            if (fast_exp(i, c_number-1, c_number) != 1):
                return False

    return True
````

````python
primes = prime_range(100_000)

carmichaels = []

for n in range(3, 100_000):
    if is_carmichael(n, primes):
        carmichaels.append(n)

print(carmichaels)
print("count =", len(carmichaels))
print("largest =", max(carmichaels))
````
