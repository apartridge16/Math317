Consider the following code to encode a text string as a binary string using ASCII.  As we did in class, we use a string of seven bits for each character.

````python
def encode(text):
    encoded_text = str()
    for letter in text:
        encoded_text = encoded_text + '{0:07b}'.format(ord(letter))
    return encoded_text

def decode(text):
    decoded_text = str()
    while len(text) > 0:
        temp = text[:7]
        text = text[7:]
        decoded_text = decoded_text + chr(int(temp,2))
    return decoded_text
````

````python
def ext_euclidean(a, b):

    s2 = 1
    s1 = 0

    t2 = 0
    t1 = 1

    r2 = a
    r1 = b

    while r1 != 0:

        q = r2 // r1

        r = r2 - q * r1
        r2 = r1
        r1 = r

        s = s2 - q * s1
        s2 = s1
        s1 = s

        t = t2 - q * t1
        t2 = t1
        t1 = t

    g = r2
    s_coeff = s2
    t_coeff = t2

    return [g, s_coeff, t_coeff]


print(ext_euclidean(a, b))
````

Use your code to find all the primitive roots modulo $101$.

````python
def is_prim_root(g, p):

    r = []

    k = 1
    while k <= p - 1:

        value = pow(g, k, p)

        if value in r:
            return False

        r.append(value)
        k = k + 1

    if len(r) == p - 1:
        return True
    else:
        return False
````

````python
def all_primitive_roots(p):
    roots = []
    g = 1
    while g <= p - 1:
        if is_prim_root(g, p):
            roots.append(g)
        g = g + 1
    return roots


print(all_primitive_roots(101))
print("Total:", len(all_primitive_roots(101)))
````

## Problem 5

Write your own `prime_range()` function to output all primes up to a given bound, using the sieve of Eratosthenes.

````python
def prime_range(n):

    sieve = [True] * (n + 1)

    sieve[0] = False
    sieve[1] = False

    p = 2
    while p * p <= n:

        if sieve[p] == True:

            multiple = p * p
            while multiple <= n:
                sieve[multiple] = False
                multiple = multiple + p

        p += 1

    primes = []
    i = 2
    while i <= n:
        if sieve[i] == True:
            primes.append(i)
        i += 1

    return primes
````
