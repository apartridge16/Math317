````python
def sbsgs_dl(g, p, h):
    n = 1 + floor(sqrt(p - 1))
    g = Mod(g, p)
    h = Mod(h, p)
    
    baby_steps = []
    for i in range(n):
        baby_steps.append(g^i)

    giant_steps = []
    for j in range(n):
        giant_steps.append(h * g^(-j*n))

    for i in range(n):
        for j in range(n):
            if baby_steps[i] == giant_steps[j]:
                y = i + j*n
                return y
````
