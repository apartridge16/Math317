```python
import math
```

```python
def gcd(a, b):
    r = 1
    while (r >= 0 and b != 0):
        q = math.floor(a / b)
        r = a % b
        if r == 0:
            break
        a = b
        b = r

    return b
```

```python
gcd(987, 707)
```

```python
def calc_units(M):
    L = list()
    for m in range(M):
        if gcd(m, M) == 1:
            L.append(str(m))

    return ', '.join(L)
```

```python
calc_units(16)
```

```python
calc_units(11)
```
