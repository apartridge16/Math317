```python
import math
```

```python
def gcd(a, b):
    r = 1
    while (r >= 0 and b != 0):
        q = math.floor(a / b)
        r = a % b
        if r == 0:
            break
        a = b
        b = r

    return b
```

```python
gcd(987, 707)
```

```python
gcd(96, 30)
```

```python
gcd(20,10)
```
